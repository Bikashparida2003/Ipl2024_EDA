```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
ipl=pd.read_csv(r"C:\Users\DELL\Desktop\ipl_data.csv")
ipl
```

```python
ipl.shape
```

```python
# getting the most man of the match awards
ipl["player_of_the_match"].value_counts()
```

```python
# getting the top 10 players with most man of the match awards
ipl["player_of_the_match"].value_counts()[0:10]
```

```python
# making a bar-plot for the top 5 players with most man of the match awards
top_players = ipl['player_of_the_match'].value_counts().head(5)

plt.figure(figsize=(8, 5))
plt.bar(top_players.index, top_players.values, color="g")
plt.xlabel('Players')
plt.ylabel('Number of Awards')
plt.title('Top 5 Players with Most "Player of the Match" Awards')
plt.show()
```

```python
# getting the frequency of winner column
ipl["winner"].value_counts()
```

```python
# finding the no of toss winner 
ipl["toss_winner"].value_counts()
```

```python
# extracting the records where a team won batting first
batting_first=ipl[ipl["win_by_runs"]!=0]
```

```python
batting_first.head()
```

```python
# making a histogram
plt.figure(figsize=(5,5))
plt.hist(batting_first["win_by_runs"])
plt.title("Distribution of Runs")
plt.xlabel("Runs")
plt.show()
```

```python
# finding the number of wins each team after batting first
batting_first["winner"].value_counts()
```

```python
# Top 5 teams with most wins after batting first
top_teams = batting_first["winner"].value_counts().head(5)

plt.figure(figsize=(7, 7))
plt.bar(top_teams.index, top_teams.values, color=["blue", "yellow", "orange","red","green"])
plt.xlabel('Teams')
plt.ylabel('Number of Wins')
plt.title('Top 5 Teams with Most Wins After Batting First')
plt.show()
```

```python
#top 10 teams wins after batting first
top_teams = batting_first["winner"].value_counts().head(10)

plt.figure(figsize=(7, 7))
plt.pie(top_teams.values, labels=top_teams.index, autopct='%1.1f%%')
plt.title('Top 10 Teams with Most Wins After Batting First')
plt.show()
```

```python
# find those records where a team has won after batting second
batting_second=ipl[ipl["win_by_wickets"]!=0]
```

```python
batting_second.head()
```

```python
# making a histogram for wins number of wickets
plt.figure(figsize=(5,5))
plt.hist(batting_second["win_by_wickets"],bins=20)
plt.title("won By wickets")
plt.xlabel("fall of wickets")
plt.show()
```

```python
# teams won batting second
batting_second["winner"].value_counts()
```

```python
# plot bar chat for teams won batting second
top_teams = batting_second["winner"].value_counts().head(5)

plt.figure(figsize=(7, 7))
plt.bar(top_teams.index, top_teams.values, color=["blue", "yellow", "orange","red","green"])
plt.xlabel('Teams')
plt.ylabel('Number of Wins')
plt.title('Top 5 Teams with Most Wins After Batting second')
plt.show()
```

```python
#top 10 teams wins after batting second
top_teams = batting_second["winner"].value_counts().head(10)

plt.figure(figsize=(7, 7))
plt.pie(top_teams.values, labels=top_teams.index, autopct='%1.1f%%')
plt.title('Top 10 Teams with Most Wins After Batting second')
plt.show()
```
